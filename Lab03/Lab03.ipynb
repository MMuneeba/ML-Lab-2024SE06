## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

### Task 3(a): Missing values

```python
print("T3a: Missing values per column (before cleaning):")
print(df.isna().sum())

rows_before = len(df)
df = df.dropna(subset=["Brand", "Model"]).copy()          # drop ads with no Brand or no Model
print("T3a: Rows removed (no Brand or no Model):", rows_before - len(df), "rows")

for col in ["Registered City", "Fuel", "Transaction Type"]:   # text columns -> "Unknown"
    df[col] = df[col].fillna("Unknown")
# Year and KMs Driven are left as NaN on purpose
print("T3a: Rows left:", len(df))
print("T3a: Missing values per column (after):")
print(df.isna().sum())
```

### Task 3(b): Formatting

```python
df["Registered City"] = df["Registered City"].str.strip().str.title()
df["Price"] = df["Price"].str.replace("Rs", "", regex=False).str.replace(",", "", regex=False).str.strip()

print("T3b: City values cleaned. Different cities now:", df["Registered City"].nunique())
print("T3b: Price examples (still text):", df["Price"].head(3).tolist(), "| dtype:", df["Price"].dtype)
```

### Task 3(c): Data types

```python
df["Price"] = df["Price"].astype("int64")
df["Year"] = df["Year"].astype("Int64")      # whole numbers that allow NaN

print("T3c: Data types after conversion:")
print(df.dtypes)
```

### Task 3(d): Inconsistent categories

```python
cities_before = df["Registered City"].nunique()
print("T3d: Different cities before:", cities_before)

df["Registered City"] = df["Registered City"].replace({"Sukkar": "Sukkur"})
df["Model"] = df["Model"].replace({"Corrolla Altis": "Corolla Altis"})

cities_after = df["Registered City"].nunique()
print("T3d: Different cities after:", cities_after)
print("T3d: 'Corrolla Altis' ads left:", (df["Model"] == "Corrolla Altis").sum())
```

### Task 3(e): Duplicates

```python
duplicate_count = df.duplicated().sum()
print("T3e: Exact duplicate ads:", duplicate_count)

df = df.drop_duplicates()
print("T3e: Rows after dropping duplicates:", len(df))
```

### Task 3(f): Invalid values

```python
mean_km_before = df["KMs Driven"].mean()
mean_price_before = df["Price"].mean() / 100_000

# Impossible kilometres -> NaN
km = df["KMs Driven"]
bad_km = (km < 100) | (km > 1_000_000) | km.isin([12345, 123456])
print("T3f: Impossible KMs Driven values set to NaN:", bad_km.sum())
df.loc[bad_km, "KMs Driven"] = np.nan

# Fake prices -> drop
fake_price = df["Price"].isin([12_345_678, 87_654_321])
print("T3f: Ads with fake prices dropped:", fake_price.sum())
df = df[~fake_price]

# Suzuki / Daihatsu above 5,000,000 -> drop
too_expensive = df["Brand"].isin(["Suzuki", "Daihatsu"]) & (df["Price"] > 5_000_000)
print("T3f: Suzuki/Daihatsu ads above 5,000,000 dropped:", too_expensive.sum())
df = df[~too_expensive].copy()

print("T3f: Rows left:", len(df))
print("T3f: Mean KMs Driven before / after:", round(mean_km_before), "/", round(df["KMs Driven"].mean()), "km")
print("T3f: Mean price before / after (lakh):", round(mean_price_before, 2), "/", round(df["Price"].mean() / 100_000, 2))
```

### Task 3(g): Self-check

```python
print("T3g: Final shape:", df.shape)
print("T3g: Matches expected (20484, 9):", df.shape == (20484, 9))
```

### Task 4(f): Corolla GLI median price for each year 2010-2018

```python
gli = df[(df["Model"] == "Corolla GLI") & (df["Year"].between(2010, 2018))]
gli_median = gli.groupby("Year")["Price_lakh"].median()
print("T4f: Corolla GLI median Price_lakh by year:")
print(gli_median)

yearly_change = gli_median.diff()
print("T4f: Change from the previous year (lakh):")
print(yearly_change.round(2))
loss_per_year = yearly_change.mean()          # average gap between neighbouring model years
print("T4f: Average value lost per year of age (lakh):", round(loss_per_year, 2))
yearly_rate = (gli_median.loc[2010] / gli_median.loc[2018]) ** (1 / 8)
print("T4f: Value kept each year (older / newer):", round(yearly_rate, 3), "-> about", round((1 - yearly_rate) * 100, 1), "% lost per year")
```

**T4f answer:** The median price of a Corolla GLI rises steadily with the model year, from 12.25 lakh (2010) to 23.20 lakh (2018). So an extra year of age costs about **1.37 lakh** on average ((23.20 − 12.25) / 8), which is roughly **7.7 % of its value per year**. The yearly gap is smaller for older cars (about 0.8 to 1.1 lakh) and bigger for newer ones (2.0 and 2.2 lakh in the last two years),this shows that the price difference between newer model years is larger, so depreciation is not constant every year.. This compares ads of different model years sold in the same period; it does not follow one car over time.

This is surprising because the installment cars are newer (median age 4 years against 13) and 44 % are New. Even the same model is cheaper (Alto 2.3 vs 4.8 lakh, Cultus VXR 2.5 vs 6.0 lakh). The most likely explanation is that an installment ad shows the down payment or an instalment amount instead of the full price, so these two medians should not be compared directly.
