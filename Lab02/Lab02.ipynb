```python
np.random.seed(42)
dice_rolls = np.random.randint(1, 7, 10000)

faces, counts = np.unique(dice_rolls, return_counts=True)

print("T3c: Dice Face Counts:", dict(zip(faces.tolist(), counts.tolist())), "rolls")
```

## Task 9 · Grace marks
Using `marks`:
- **(a)** Add 5 grace marks but cap at 100 (`np.clip`).
- **(b)** Pass % (≥ 50) after grace.
- **(c)** Index of students who passed **only because of** the grace marks.
- **(d)** z-score of each student after grace.

### Task 9(a): Add 5 grace marks and cap at 100


```python
final_marks = np.clip(marks + 5, 0, 100)

print("T9a: Final Marks After Grace:", final_marks, "marks")
```

### Task 9(b): Calculate pass percentage (passing marks = 50)


```python
passed = final_marks >= 50

pass_percentage = np.mean(passed) * 100

print("T9b: Pass Percentage:", round(pass_percentage, 2), "%")
```

### Task 9(c): Students who passed only because of grace marks


```python
grace_only_passed = (marks < 50) & (final_marks >= 50)
print("T9c: Indices of students who passed only because of grace:", np.where(grace_only_passed)[0])
```

### Task 9(d): Calculate z-scores


```python
z_scores = (final_marks - np.mean(final_marks)) / np.std(final_marks)

print("T9d: Z-Scores after grace:", np.round(z_scores, 2))
```

### Task 10(a): Create an 8x8 canvas and draw number 7


```python
# Create an 8x8 black canvas
img = np.zeros((8, 8), dtype=int)

# Draw top horizontal line
img[1, 1:7] = 255

# Draw diagonal line
img[2, 6] = 255
img[3, 5] = 255
img[4, 4] = 255
img[5, 3] = 255
img[6, 2] = 255

# Add grey pixels along the diagonal
img[2, 5] = 128
img[3, 4] = 128
img[4, 3] = 128
img[5, 2] = 128
img[6, 1] = 128

# Display array
print("T10a: Original Image Array:")
print(img)

# Display image
plt.figure(figsize=(5, 5))
plt.imshow(img, cmap='gray', vmin=0, vmax=255)
plt.title("Classic Digit 7")
plt.axis('off')
plt.show()
```

### Task 10(b): Flip the image Left-Right


```python
flipped_img = np.fliplr(img)

print("T10b: Flipped Image Array:")
print(flipped_img)

plt.figure(figsize=(5, 5))
plt.imshow(flipped_img, cmap='gray', vmin=0, vmax=255)
plt.title("Flipped Image")
plt.axis('off')
plt.show()
```

### Task 10(c): Invert image colors



```python
inverted_img = 255 - img

print("T10c: Inverted Image Array:")
print(inverted_img)

plt.figure(figsize=(5, 5))
plt.imshow(inverted_img, cmap='gray', vmin=0, vmax=255)
plt.title("Inverted Image")
plt.axis('off')
plt.show()
```

### Task 10(d):Crop the image


```python
cropped_img = img[1:5, 3:7]

print("T10d: Cropped Image Array:")
print(cropped_img)

plt.figure(figsize=(5, 5))
plt.imshow(cropped_img, cmap='gray', vmin=0, vmax=255)
plt.title("Cropped Image")
plt.axis('off')
plt.show()
```

### Task 10(e): Brighten image with a maximum cap of 255


```python
bright_img = np.clip(img + 80, 0, 255)

print("T10e: Brightened Image Array:")
print(bright_img)

plt.figure(figsize=(5, 5))
plt.imshow(bright_img, cmap='gray', vmin=0, vmax=255)
plt.title("Brightened Image")
plt.axis('off')
plt.show()
```

### Task 10(f): Shape, data type and white pixels


```python
print("T10f: Image Shape:", img.shape)
print("T10f: Image Data Type:", img.dtype)

white_pixels = np.sum(img == 255)

print("T10f: Number of White Pixels:", white_pixels)
```
